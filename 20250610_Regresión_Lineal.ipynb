```python
# Importar librería
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('PrecioMetroCuadrado.csv')
x = dataset.iloc[ : , :-2 ]
y = dataset.iloc[ : , -1 ]
#y
```

```python
# train_test_split: dividir el dataset en 4 variables
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 0.2, random_state=0)
#x_test.shape
```

```python
# Librería scikit-learn para Machine Learning
from sklearn.linear_model import LinearRegression
# Obtener una instancia
model = LinearRegression()
# Fit = Entrenar el modelo
model.fit(x_train, y_train)
```

```python
#Intentar predecir todo el arreglo x_test
ypred = model.predict(x_test)
ypred
```

```python
#y_test
```

```python
# Diagrama de disperción: intensidad de la relación entre dos variables numéricas
plt.scatter(x_train, y_train, color="red")
plt.plot(x_train, model.predict(x_train), color="blue")
```

```python
# MÉTRICAS
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
print("R2:   ", r2_score(y_test, ypred))
print("MAE:  ", mean_absolute_error(y_test, ypred))
print("MSE:  ", mean_squared_error(y_test, ypred))
print("RMSE: ", np.sqrt(mean_squared_error(y_test, ypred)))
```
